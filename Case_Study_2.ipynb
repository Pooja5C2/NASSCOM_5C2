# **Part F – Calculus: Differentiation**

Task 6:

20. Find dF/dx manually.
21. Show that F′(x) = 5 − 0.4x.
22. Find the value of x for which F′(x) = 0.
23. Interpret the result in the context of study hours.

Given:
F(x) = 40 + 5x - 0.2x^2 $$

The derivative is:
F′(x)=5−0.4x

```python
def F(x):
    return 40 + 5*x - 0.2*x*x

def dF(x):
    return 5 - 0.4*x

for x in range(1, 11):
    print(x, F(x), dF(x))

# To find where the derivative is zero:
    x = 5 / 0.4
print("x =", x)
```

```python
x = np.linspace(0, 5, 100)
y = 5*x - 0.2*x*x
area = np.trapezoid(y, x)
print("Approximate area =", area)
```

# **Part I – Mini Prediction Model Using Linear Algebra**

```python
X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values
Y = df["Final_Score"].values
X = np.column_stack((np.ones(len(X)), X))
beta = np.linalg.inv(X.T @ X) @ X.T @ Y
print("Coefficients:")
print(beta)
```
