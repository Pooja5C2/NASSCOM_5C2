Import NumPy and Store Battery Data

```python
import numpy as np

# Battery sensor data
battery_data = [
    [0, 32, 400, 20, 95],
    [1, 35, 395, 22, 88],
    [2, 38, 390, 25, 80],
    [3, 40, 385, 28, 70],
    [4, 42, 380, 30, 60],
    [5, 45, 375, 32, 48]
]

print("Battery Sensor Data")
print("-------------------")

for reading in battery_data:
    print("Time:", reading[0], "h",
          "| Temperature:", reading[1], "°C",
          "| Voltage:", reading[2], "V",
          "| Current:", reading[3], "A",
          "| Charge:", reading[4], "%")
```

High Temperature Alert

```python
print("Temperature Status")
print("------------------")

for reading in battery_data:
    temperature = reading[1]

    if temperature > 40:
        print("Time:", reading[0], "h - High Temperature Alert")
    else:
        print("Time:", reading[0], "h - Temperature Normal")
```

Low Battery Alert

```python
print("Battery Charge Status")
print("---------------------")

for reading in battery_data:
    charge = reading[4]

    if charge < 50:
        print("Time:", reading[0], "h - Low Battery Alert")
    else:
        print("Time:", reading[0], "h - Battery Charge Normal")
```

Battery Status Function

```python
def battery_status(charge):
    if charge > 70:
        return "High"
    elif charge >= 50:
        return "Medium"
    else:
        return "Low"


print("Battery Status")
print("--------------")

for reading in battery_data:
    time = reading[0]
    charge = reading[4]

    print("Time:", time, "h | Charge:", charge, "% | Status:", battery_status(charge))
```

Convert Data into NumPy Array

```python
D = np.array([
    [32, 400, 20, 95],
    [35, 395, 22, 88],
    [38, 390, 25, 80],
    [40, 385, 28, 70],
    [42, 380, 30, 60],
    [45, 375, 32, 48]
])

print("Battery Data Matrix:")
print(D)
```

Shape of the Array

```python
print("Shape of the array:", D.shape)
print("Number of rows:", D.shape[0])
print("Number of columns:", D.shape[1])
```

Average of Each Parameter

```python
average = np.mean(D, axis=0)

print("Average Temperature:", average[0], "°C")
print("Average Voltage:", average[1], "V")
print("Average Current:", average[2], "A")
print("Average Charge:", average[3], "%")
```

Maximum and Minimum Temperature

```python
max_temperature = np.max(D[:, 0])
min_temperature = np.min(D[:, 0])

print("Maximum Temperature:", max_temperature, "°C")
print("Minimum Temperature:", min_temperature, "°C")
```

Extract Voltage Column

```python
voltage = D[:, 1]

print("Voltage Column:")
print(voltage)
```

Identify High-Temperature Readings

```python
high_temperature = D[D[:, 0] > 40]

print("High-Temperature Readings:")
print(high_temperature)
```

Identify Low-Battery Readings

```python
low_battery = D[D[:, 3] < 50]

print("Low-Battery Readings:")
print(low_battery)
```

Task 3 — Linear Algebra
Define Matrix A and Vector X

```python
A = np.array([
    [2, 1],
    [1, 3]
])

X = np.array([
    40,
    25
])

print("Matrix A:")
print(A)

print("\nVector X:")
print(X)
```

Calculate Y = AX

```python
Y = A @ X

print("Y = AX")
print("Y =", Y)
```

Transpose of A

```python
A_transpose = A.T

print("Transpose of A:")
print(A_transpose)
```

Determinant of A

```python
det_A = np.linalg.det(A)

print("Determinant of A:", det_A)
```

Inverse of A

```python
A_inverse = np.linalg.inv(A)

print("Inverse of A:")
print(A_inverse)
```

Solve the System of Equations

```python
B = np.array([
    100,
    120
])

solution = np.linalg.solve(A, B)

x = solution[0]
y = solution[1]

print("Solution:")
print("x =", x)
print("y =", y)
```

Verify the Solution

```python
print("Verification:")

print("2x + y =", 2*x + y)
print("x + 3y =", x + 3*y)
```

Task 4 — Calculus
Define Charge Function

```python
def charge_function(t):
    return 100 - 2*t - 0.5*t**2

print("Charge at t = 0:", charge_function(0), "%")
print("Charge at t = 2:", charge_function(2), "%")
print("Charge at t = 5:", charge_function(5), "%")
```

Derivative of Charge Function

```python
def discharge_rate(t):
    return -2 - t

print("dC/dt =", "-2 - t")
```

Discharge Rate at t = 2

```python
rate_at_2 = discharge_rate(2)

print("Discharge rate at t = 2 hours:", rate_at_2, "% per hour")
```

Discharge Rate at t = 5

```python
rate_at_5 = discharge_rate(5)

print("Discharge rate at t = 5 hours:", rate_at_5, "% per hour")
```

Calculus Interpretation

```python
if abs(rate_at_5) > abs(rate_at_2):
    print("The battery is losing charge at an increasing rate.")
elif abs(rate_at_5) == abs(rate_at_2):
    print("The battery is losing charge at a constant rate.")
else:
    print("The battery is losing charge at a decreasing rate.")
```
